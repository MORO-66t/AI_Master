# Task 8.
- Write a Python program to find all the unique words and count the frequency of occurrence from a text file and save the counts in a new file
- This is the same as Task 7 but this time read words from a file
- Hint: use replace function and split function

**First Step: read the file**

```python

input_file = open("input_file.txt", "r")
text = input_file.read()
input_file.close()
```

**Second step: remove '\n' then convert the file text into a list of words**

```python
text = text.replace('\n', ' ')
words = text.split(' ')
```

**Third Step: get unique words, create a dictionary, then count unique words**

```python
unique_words = set(words)
word_counts = dict()

for word in unique_words:
    word_counts[word] = words.count(word)
```

**Save the word-count pair into a text file**

```python
output_file = open("output_file.txt", "w")
for word in word_counts:
    output_file.write("{}: {}\n".format(word, word_counts[word]))
output_file.close()
```

```python
### convert first to function
#### your class here

class WordsCounts:
    """
    WordsCounts class handles reading words from a file, counting the frequency
    of each unique word, and saving the counts to an output file.

    Attributes:
        filename (str): The name of the input file containing the text.
    """

    def __init__(self, filename):
        """
        Initializes a WordsCounts object with the input file name.

        Args:
            filename (str): The name of the input file containing the text.
        """
        self.filename = filename

    def openfile(self, filename):
        """
        Reads the contents of a file and returns them as a string.

        Args:
            filename (str): The name of the file to read.

        Returns:
            str: The full text content of the file.
        """
        with open(filename, "r") as f:
            text = f.read()
            return text

    def words(self, text):
        """
        Replaces newline characters with spaces.

        Args:
            text (str): The text to process.

        Returns:
            str: The text with newline characters replaced by spaces.
        """
        text = text.replace('\n', ' ')
        text.split(' ')
        return text

    def unique_words(self, text):
        """
        Extracts the set of unique words present in the text.

        Args:
            text (str): The text to process.

        Returns:
            set: A set containing the unique words of the text.
        """
        words = text.split()
        unique_words = set(words)
        return unique_words

    def count_words(self, words, unique_words):
        """
        Counts how many times each unique word appears in the words list.

        Args:
            words (list): The full list of words.
            unique_words (set): The set of unique words to count.

        Returns:
            dict: A dictionary mapping each unique word to its frequency.
        """
        word_counts = {}
        for word in unique_words:
            word_counts[word] = words.count(word)
        return word_counts

    def savefile(self, filename, word_counts):
        """
        Writes the word-count pairs into an output file.

        Args:
            filename (str): The name of the file to write to.
            word_counts (dict): A dictionary mapping each word to its frequency.
        """
        with open(filename, "w") as f:
            for word, count in word_counts.items():
                f.write(f"{word}: {count}\n")


#### test your class
if __name__ == "__main__":
    w = WordsCounts("input_file.txt")
    text = w.openfile("input_file.txt")
    uc = w.unique_words(text)
    words = w.words(text)
    c = w.count_words(words, uc)
    w.savefile("outpot.txt", c)
```

# Task 9.
- Implement a basic chatbot that can respond to user input with predefined responses

```python
import random

# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}

# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])

    return random.choice(responses["default"])

# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")

    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)

        if user_input == "goodbye":
            break

# Run the chatbot
chatbot()
```

```python
#### your class here

import random

class Chatbot:
    """
    Chatbot class handles responding to user input 

    Attributes:
        responses (dict): A dictionary mapping keywords to lists of replies
    """

    def __init__(self, responses):
        """
        Initializes the Chatbot

        Args:
            responses (dict): A dictionary where each key is a keyword and each
                              value is a list of possible responses for that keyword.
        """
        self.responses = responses

    def get_response(self, user_input):
        """
        Determines an appropriate reply based on the user's input.

        Args:
            user_input (str): The text entered by the user.

        Returns:
            str: A random response matching a keyword in the user's input,
                 or a random default response if no keyword matches.
        """
        for key in self.responses:
            if key in user_input:
                return random.choice(self.responses[key])

        return random.choice(self.responses["default"])

    def run(self):
        """
        Runs the interactive chatbot loop.

        Greets the user, repeatedly reads input, prints a response and stops
        when the user types 'goodbye'.
        """
        print("Chatbot: Hi! How can I assist you today?")

        while True:
            user_input = input("User: ").lower()
            response = self.get_response(user_input)
            print("Chatbot:", response)

            if user_input == "goodbye":
                break


#### test your class
if __name__ == "__main__":
    responses = {
        "hello": ["Hello!", "Hi there!", "Greetings!"],
        "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
        "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
        "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
    }

    bot = Chatbot(responses)
    bot.run()
```

# Task 10: Person class and child Student class
Create a Python class Person with attributes: name and age.
- Create a display() method that displays the name and age of an object created via the Person class.
- Create a child class Student which inherits from the Person class and which also has an extra attribute called school.
- Create a method displayStudent() that displays the name, age and section of an object created via the Student class.
- Create a student object via an instantiation on the Student class and then test the displayStudent method

```python
class Person:
    """
    Person class represents a person with a name and an age

    Attributes:
        name (str): The person's name
        age (int): The person's age
    """

    def __init__(self, name, age):
        """
        Initializes a Person object

        Args:
            name (str): The person's name
            age (int): The person's age
        """
        self.name = name
        self.age = age

    def display(self):
        """
        Displays the name and the age of the person

        Returns:
            None: This method only prints the person's name and age
        """
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):
    """
    Student class inherits from Person and adds a school attribute.

    Attributes:
        name (str): The student's name.
        age (int): The student's age.
        school (str): The school the student attends.
    """

    def __init__(self, name, age, school):
        """
        Initializes a Student object.

        Args:
            name (str): The student's name.
            age (int): The student's age.
            school (str): The school the student attends.
        """
        super().__init__(name, age)
        self.school = school

    def displayStudent(self):
        """
        Displays the name, age and school of the student.

        Returns:
            None: This method only prints the student's name, age and school.
        """
        self.display()
        print("School:", self.school)


# Create a student object via instantiation on the Student class
student1 = Student("Ali", 20, "Modern School")
student1.displayStudent()
```
