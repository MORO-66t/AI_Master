# Task 1.
- Calculate Factorial using recursion

```python
def factorial(num : int) -> int:
    '''
    factorial: calculate the factorial of a given number using recursion

    :param num: the number to calculate its factorial
    :type num: integer

    :return: factorial of num
    '''
    if num == 0:
        return 0
    if num == 1:
        return 1
    return num * factorial(num - 1)

factorial(5)
```

# Task 2.
- Write a function that takes a number and return True if this number is prime and False if not.

```python
def isPrime(num : int) -> bool:
    '''
    isPrime: check if a given number is prime

    :param num: the number to check
    :type num: integer

    :return: True if the number is prime, False otherwise
    '''
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

isPrime(20)
```

# Task 3.
- Write a function that takes two numbers and return a list of common dividors

```python
def dividors(num1 : int, num2 : int) -> list:
    '''
    dividors: return a list of the common divisors of two numbers

    :param num1: the first number
    :type num1: integer

    :param num2: the second number
    :type num2: integer

    :return: list of the common divisors
    '''
    mn = min(num1, num2)
    common_dividors = []
    for i in range(1, mn + 1):
        if num1 % i == 0 and num2 % i == 0:
            common_dividors.append(i)
    return common_dividors

dividors(20, 10)
```

# Task 4. 
- Given two strings, write a program that efficiently finds the longest common subsequence

- **Ex:** str1 = "Welcome to Machine Learning Diploma" str2 = "I am studying a Machine Learning Course" --> Machine Learning

```python
def longest_common_substring(str1 : str, str2 : str) -> str:
    '''
    longest_common_substring: find the longest common substring between two strings

    :param str1: the first string
    :type str1: string

    :param str2: the second string
    :type str2: string

    :return: the longest common substring
    '''
    longest_str = ""
    for i in range(len(str1)):
        for j in range(i, len(str1)):
            sub_str = str1[i:j + 1]
            if sub_str in str2 and len(sub_str) > len(longest_str):
                longest_str = sub_str
    return longest_str

str1 = "Welcome to Machine Learning Diploma"
str2 = "I am studying a Machine Learning Course"
longest_common_substring(str1, str2)
```

# Task 5.
- Write a function that will take a given string and reverse the order of words.

- **Ex:** “Hello world” → ”world Hello”


```python
def reverse_words(s : str) -> str:
    '''
    reverse_words: reverse the order of the words in a given string

    :param s: the input string
    :type s: string

    :return: the string with its words in reversed order
    '''
    lst = s.split(" ")
    new_lst = lst[::-1]
    return " ".join(new_lst)

reverse_words("Hello World . to the ")
```

# Task 6.
- Create a program that generates a random password of a given length.
- Hint: use random module → import random → random.randrange

```python
import random

def generate_password(length : int) -> str:
    '''
    generate_password: generate a random password of a given length

    :param length: the length of the password
    :type length: integer

    :return: a random password of the given length
    '''
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
    num_of_chars = len(characters)
    password = ""
    for _ in range(length):
        rand_indx = random.randrange(0, num_of_chars)
        password += characters[rand_indx]
    return password

password_length = 8
random_password = generate_password(password_length)
print(random_password)
```

# Task 7.
- Write a Python program to find all the unique words and count the frequency of occurrence from a given list of strings

- **Ex:** ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]  → {'Ali': 4, 'Welcome': 1, 'No': 3, 'Hi': 2}

```python
def word_frequency(words : list) -> dict:
    '''
    word_frequency: count the frequency of every unique word in a list of strings

    :param words: a list of strings
    :type words: list

    :return: a dictionary with the unique words and their frequencies
    '''
    unique_words = set(words)
    counts = {word: 0 for word in unique_words}
    for word in unique_words:
        for item in words:
            if word == item:
                counts[word] += 1
    return counts

words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]
word_frequency(words)
```
