```python
def multiplication_table(num : int) -> None:
    '''
    multiplication_table: print the multiplication table of a given number

    :param num: the number to print its multiplication table
    :type num: integer

    :return: None (prints the multiplication table)
    '''
    for i in range(1, 13):
        print(f"{num} x {i} = {num * i}")

multiplication_table(5)
```

```python
def isPrime(num : int) -> bool:
    '''
    isPrime: check if a given number is prime

    :param num: the number to check
    :type num: integer

    :return: True if the number is prime, False otherwise
    '''
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True

def twin_primes(limit : int) -> list:
    '''
    twin_primes: find all pairs of twin primes less than a given limit
    (two consecutive odd numbers that are both prime)

    :param limit: the upper limit (exclusive)
    :type limit: integer

    :return: list of twin-prime pairs
    '''
    twins = []
    for num in range(2, limit):
        if isPrime(num) and isPrime(num + 2):
            twins.append((num, num + 2))
    return twins

twin_primes(1000)
```

```python
def prime_factors(num : int) -> list:
    '''
    prime_factors: find the prime factors of a given number

    :param num: the number to find its prime factors
    :type num: integer

    :return: list of the prime factors of num
    '''
    factors = []
    divisor = 2
    while divisor * divisor <= num:
        while num % divisor == 0:
            factors.append(divisor)
            num //= divisor
        divisor += 1
    if num > 1:
        factors.append(num)
    return factors

prime_factors(56)
```

```python
def decimal_to_binary(num : int) -> str:
    '''
    decimal_to_binary: convert a decimal number to its binary representation

    :param num: the decimal number to convert
    :type num: integer

    :return: the binary representation of num as a string
    '''
    if num == 0:
        return "0"
    binary = ""
    while num > 0:
        binary = str(num % 2) + binary
        num //= 2
    return binary

decimal_to_binary(10)
```

```python
def is_perfect(num : int) -> bool:
    '''
    is_perfect: check if a number is perfect (sum of proper divisors equals the number)

    :param num: the number to check
    :type num: integer

    :return: True if the number is perfect, False otherwise
    '''
    if num < 2:
        return False
    total = 0
    for i in range(1, num):
        if num % i == 0:
            total += i
    return total == num

def perfect_numbers(start : int, end : int) -> list:
    '''
    perfect_numbers: print all the perfect numbers in a given range

    :param start: the start of the range
    :type start: integer

    :param end: the end of the range
    :type end: integer

    :return: list of the perfect numbers in the range
    '''
    perfect = []
    for num in range(start, end + 1):
        if is_perfect(num):
            perfect.append(num)
    return perfect

perfect_numbers(1, 1000)
```
